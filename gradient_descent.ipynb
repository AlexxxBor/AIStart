Сначала нарисуем график функции f(x) = x<sup>4</sup> - 10x + 3:

```python
import numpy as np
from matplotlib import pyplot as plt

def f(x):
    return x ** 4 - 10 * x + 3

x_values = [x for x in np.arange(-5, 5, 0.1)]
f_values = [f(x) for x in np.arange(-5, 5, 0.1)]

plt.figure(figsize=(8, 8))

plt.axvline(x=0, c='black')
plt.axhline(y=0, c='black')

plt.plot(x_values, f_values)
```

Вычисляем производную функции f(x) и используем её для реализации градиентного спуска:


```python
def grad_f(x):
    return 4 * x ** 3 - 10

def gradient_descent(x_start, learning_rate, epsilon, iterations_count):
    x_current = x_start

    for _ in range(iterations_count):
       x_new = x_current - learning_rate * grad_f(x_current)

       if abs(x_new - x_current) < epsilon:
           return x_current 

       x_current = x_new
       
    return x_current
```

Найдём локальный минимум при значениях:
- x<sub>0</sub> = 5
- η = 0.01
- epsilon = 0.001
- iterations = 10000

```python
x0 = 5
n = 0.01
epsilon = 0.001
num_iterations = 10000

x_result = gradient_descent(x0, n, epsilon, num_iterations)
f'{x_result:.2f}'
```

Отобразим точку на графике:

```python
plt.figure(figsize=(8, 8))

plt.axvline(x=0, c='black')
plt.axhline(y=0, c='black')
plt.plot(x_values, f_values)

plt.scatter(x_result, f(x_result), marker='o', c='red')
```
